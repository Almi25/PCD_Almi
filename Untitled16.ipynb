```python
import imageio.v3 as iio
import numpy as np
from scipy import ndimage
import matplotlib.pyplot as plt

img = iio.imread("/content/fullmoon.jpg")
if img.ndim == 3:
    img = np.dot(img[..., :3], [0.299, 0.587, 0.114])
img = img.astype(np.uint8)

# Ekualisasi histogram hanya pada piksel bulan (abaikan latar hitam)
mask = img > 15
hist, _ = np.histogram(img[mask], bins=256, range=(0, 256))
cdf = hist.cumsum()
cdf_min = cdf[cdf > 0].min()
lut = ((cdf - cdf_min) / (cdf.max() - cdf_min) * 255).clip(0, 255).astype(np.uint8)

equalized = img.copy()
equalized[mask] = lut[img[mask]]

# Penghalusan & penajaman dengan scipy.ndimage
smooth = ndimage.gaussian_filter(equalized.astype(float), sigma=1)
sharp = np.clip(equalized + 1.5 * (equalized - smooth), 0, 255).astype(np.uint8)

iio.imwrite("hasil.png", sharp)

# Tampilkan citra dan histogramnya
fig, ax = plt.subplots(2, 3, figsize=(14, 8))
for i, (im, t) in enumerate(zip([img, equalized, sharp],
                                ["Asli", "Equalized", "Equalized + Sharpen"])):
    ax[0, i].imshow(im, cmap="gray", vmin=0, vmax=255)
    ax[0, i].set_title(t)
    ax[0, i].axis("off")
    ax[1, i].hist(im[im > 15].ravel(), bins=256, range=(0, 256), color="gray")
    ax[1, i].set_title(f"Histogram {t}")
plt.tight_layout()
plt.show()
```
